```python
from sklearn.preprocessing import LabelEncoder
# Drop unnecessary columns
columns_to_drop = [
    'policy_number', 'insured_zip', 'policy_bind_date', 'incident_date',
    'insured_hobbies', 'incident_city', 'incident_location', 'auto_model'
]
df.drop(columns_to_drop, axis=1, inplace=True, errors='ignore')

# Select final features
X = df[features_to_include].copy()
y = df[target]

# Encode categorical variables
categorical_cols = X.select_dtypes(include=['object']).columns
le_dict = {}
for col in categorical_cols:
    le = LabelEncoder()
    X[col] = le.fit_transform(X[col].astype(str))
    le_dict[col] = le

# Encode target
y = LabelEncoder().fit_transform(y)
```

```python
# For columns with natural order, use manual mapping
# For categorical without order, use one-hot or leave as label encoded

# Police report - has natural order (YES > UNKNOWN > NO in terms of documentation quality)
police_report_map = {'YES': 2, 'Unknown': 1, 'NO': 0}
df['police_report_available'] = df['police_report_available'].map(police_report_map)

# Property damage - similar approach
property_damage_map = {'YES': 2, 'Unknown': 1, 'NO': 0}
df['property_damage'] = df['property_damage'].map(property_damage_map)

# For collision_type (no natural order), use label encoding or one-hot
collision_le = LabelEncoder()
df['collision_type'] = collision_le.fit_transform(df['collision_type'])
```
