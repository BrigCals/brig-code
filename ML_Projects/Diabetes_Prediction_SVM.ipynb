```python
import pandas as pd
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn import svm
from sklearn.metrics import accuracy_score
```

```python
data = pd.read_csv(r"C:\Users\USER\Downloads\diabetes.csv")
```

```python
data_to_keep = ['Pregnancies', 'Glucose', 'BloodPressure', 'SkinThickness', 'Insulin',
       'BMI', 'DiabetesPedigreeFunction', 'Age', 'Outcome'] 
```

```python
data
```

```python
# 0 Non-diabetic
# 1 Diabetic

X = data.drop('Outcome' , axis = 1)
Y = data['Outcome']

scaler = StandardScaler()

scaler.fit(X)

standardized_data = scaler.transform(X)

X = standardized_data
```

```python
print(X)
```

```python
X_train , X_test, Y_train, Y_test = train_test_split(X, Y, test_size = 0.2, stratify = Y , random_state = 2)

print(X.shape, X_train.shape, X_test.shape)
```

```python
classifier = svm.SVC(kernel = 'linear')

classifier.fit(X_train, Y_train)

X_train_prediction  = classifier.predict(X_train) #Training data
data_training_accuracy = accuracy_score(X_train_prediction, Y_train)

X_test_prediction = classifier.predict(X_test)  #Testing data
testing_data_accuracy = accuracy_score(X_test_prediction, Y_test)

print("The accuracy score of the trained data is ", data_training_accuracy)
print("The accuracy score of the tested data is ", testing_data_accuracy)
```
